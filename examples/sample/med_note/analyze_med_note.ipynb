# Medical Note Dataset Analysis (Full)

This notebook downloads and loads the full dataset `clinical_cases.csv.gz` from Hugging Face.

```python
import os
import requests
import gzip
import shutil
import pandas as pd
import random
from pprint import pprint

# Define paths and URL
dataset_url = "https://huggingface.co/datasets/bavehackathon/2026-healthcare-ai/resolve/main/clinical_cases.csv.gz"
local_filename = "sample/med_note/clinical_cases.csv.gz"

# Ensure directory exists
os.makedirs(os.path.dirname(local_filename), exist_ok=True)

# Download the file if it doesn't exist
if not os.path.exists(local_filename):
    print(f"Downloading full dataset from {dataset_url}...")
    response = requests.get(dataset_url, stream=True)
    if response.status_code == 200:
        with open(local_filename, 'wb') as f:
            f.write(response.content)
        print("Download complete.")
    else:
        print(f"Failed to download. Status code: {response.status_code}")
else:
    print(f"File already exists: {local_filename}")
```

```python
# Load full dataset with pandas
if os.path.exists(local_filename):
    try:
        df = pd.read_csv(local_filename, compression='gzip')
        print(f"Successfully loaded {len(df)} rows.")
    except Exception as e:
        print(f"Error loading CSV: {e}")
else:
        print("File not found. Cannot load data.")
```

```python
# Print Header (Column names)
if 'df' in locals() and notdf.empty:
    print("Header (Data Columns):")
    print(list(df.columns))
else:
    print("No data loaded.")
```

```python
# Print 1 Random Entry
if 'df' in locals() and not df.empty:
    random_index = random.randint(0, len(df) - 1)
    # Convert row to dictionary for cleaner printing
    random_entry = df.iloc[random_index].to_dict()
    print("\nRandom Entry (Row index {}):".format(random_index))
    pprint(random_entry)
else:
    print("No data to display.")
```
